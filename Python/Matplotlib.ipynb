### Matplotlib :
### It is popular python library used for data visualizations . It helps to create static , animated and interactive graphs in python.

```python
import matplotlib.pyplot as plt
```

#### 1. Line Chart:
#### It is used for show the trends over the time.

```python
x = [3,7,5,9]
plt.plot(x, marker = 'o')
plt.title("Line Chart")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### Bar Chart/graph :
### It is used for comapring quatities by categories.

```python
categories = ['A' , 'B', 'C' , 'D'] # x
value= [35,55,40,60] # y

##create a bar plot
plt.bar(categories,value , color = 'skyblue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot')
plt.show()
```

```python
x=[1,5,7,8,9]
y=[1,4,9,16,25]
plt.barh(x,y,color = 'green')
plt.show()
```

```python
subs = ['C++' , 'Java' , 'HTML' , 'Pyhton']
x = [25,40,15,20]

plt.pie(x , labels = subs , autopct = '%1.f1%%' , startangle = 90)
plt.show()
```
