```python
df["annual income (k$)"] =df["annual income (k$)"].fillna("unknown values")   #for filling with constant
```

```python
df["salary"] = df["salary"].fillna()(df["salary"].mean())    #numerical and data should not contain any outliers
```

```python
df["salary"] = df["salary"].fillna()(df["salary"].median())  #numerical and data should contain  outliers
```

```python
df["salary"] = df["salary"].fillna()(df["salary"].mode())    #categorical and text kind of data
```
